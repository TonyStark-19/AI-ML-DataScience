### Load the libraries needed

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

### Load the data

```python
df = pd.read_csv("loan_approval_data.csv")
df.head()
```

### Information about data

```python
# information about the data
df.info()
```

```python
# describe function to get numerical info about data
df.describe()
```

```python
# check number of null values
df.isnull().sum()
```

### Handle Missing Values

```python
# categorical values
categorical_cols = df.select_dtypes(include=object).columns
categorical_cols
```

```python
# Remove Applicant_ID
df = df.drop("Applicant_ID", axis = 1)
```

```python
# numerical values
numerical_cols = df.select_dtypes(include=float).columns
numerical_cols
```

```python
# import simple imputer
from sklearn.impute import SimpleImputer
```

```python
# numerical imputer
num_imp = SimpleImputer(strategy="mean")
df[numerical_cols] = num_imp.fit_transform(df[numerical_cols])

# re-check the number of null values in numerical columns
df[numerical_cols].isnull().sum()
```

```python
# categorical imputer
cat_imp = SimpleImputer(strategy="most_frequent")
df[categorical_cols] = cat_imp.fit_transform(df[categorical_cols])

# re-check the number of null values in categorical columns
df[categorical_cols].isnull().sum()
```

```python
# final check of null values
df.isnull().sum()
```
