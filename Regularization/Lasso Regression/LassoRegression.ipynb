### Load the libraries needed

```python
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_squared_error, r2_score
import pandas as pd
```

### Load the Data

```python
df = pd.read_csv("insurance.csv")
df.head()
```

### Data Preprocessing

```python
# mapping sex column to numeric values
df["sex"] = df["sex"].map({"female": 1, "male": 0})
# mapping smoker column to numeric values
df["smoker"] = df["smoker"].map({"yes": 1, "no": 0})
df.head()
```

```python
# interaction feautres
df["age_smoker"] = df["age"] * df["smoker"]
df["bmi_smoker"] = df["bmi"] * df["smoker"]
df.head()
```

### Split the data into **dependent** and **independent** variables

```python
# Drop charges and region column
x = df.drop(columns=["charges", "region"], axis=1)
# independent
x.head()
```

```python
# dependent
y = df.charges
y.head()
```

### Split the data into **training** and **testing** part

```python
# import train test split
from sklearn.model_selection import train_test_split
```

```python
train_x, test_x, train_y, test_y = train_test_split(x, y, test_size=0.2, random_state = 1)
```

```python
len(train_x), len(test_x), len(train_y), len(test_y)
```

### Create Model

```python
# Create object
lasso_model = Lasso(alpha=0.5)
```

```python
# Pass the values to train the model
lasso_model.fit(train_x, train_y)
```

```python
# predict the values from the model
y_pred = lasso_model.predict(test_x)
```

```python
# calculate mean_squared_error
mse = mean_squared_error(test_y, y_pred)
print("MSE: ", mse)
```

### LassoCV

```python
# import LassoCV
from sklearn.linear_model import LassoCV
```

```python
# alpha values list
alphas = [0.001, 0.1, 1, 2, 5, 10, 20, 30, 40, 50, 100]

lasso_cv_model = LassoCV(
    alphas=alphas,
    cv=5,
    max_iter=1000,
    random_state=42
)

lasso_cv_model.fit(train_x, train_y)
```

```python
# best alpha value
print("Best alpha: ", lasso_cv_model.alpha_)
```

```python
# predict values from the model
y_pred = lasso_cv_model.predict(test_x)
```

```python
# calculate mean_squared_error
mse = mean_squared_error(test_y, y_pred)
print("MSE: ", mse)
```

```python
# calculate r2 score
r2 = r2_score(test_y, y_pred)
print("r2 score: ", r2)
```
