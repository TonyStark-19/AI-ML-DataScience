### Load the libraries needed

```python
from sklearn.linear_model import ElasticNet
from sklearn.metrics import mean_squared_error, r2_score
import pandas as pd
```

### Load the Data

```python
df = pd.read_csv("insurance.csv")
df.head()
```

### Data Preprocessing

```python
# mapping sex column to numeric values
df["sex"] = df["sex"].map({"female": 1, "male": 0})
# mapping smoker column to numeric values
df["smoker"] = df["smoker"].map({"yes": 1, "no": 0})
df.head()
```

```python
# interaction feautres
df["age_smoker"] = df["age"] * df["smoker"]
df["bmi_smoker"] = df["bmi"] * df["smoker"]
df.head()
```

### Split the data into **dependent** and **independent** variables

```python
# Drop charges and region column
x = df.drop(columns=["charges", "region"], axis=1)
# independent
x.head()
```

```python
# dependent
y = df.charges
y.head()
```

### Split the data into **training** and **testing** part

```python
# import train test split
from sklearn.model_selection import train_test_split
```

```python
train_x, test_x, train_y, test_y = train_test_split(x, y, test_size=0.2, random_state = 1)
```

```python
len(train_x), len(test_x), len(train_y), len(test_y)
```

### Create Model

```python
# Create object (alpha is total penalty, l1_ratio is the mix)
en_model = ElasticNet(alpha=0.5, l1_ratio=0.5)
```

```python
# Train the model
en_model.fit(train_x, train_y)
```

```python
# Predict values
y_pred = en_model.predict(test_x)
```

```python
# calculate mean_squared_error
mse = mean_squared_error(test_y, y_pred)
print("Elastic Net MSE: ", mse)
```

### ElasticNetCV

```python
# import ElasticNetCV
from sklearn.linear_model import ElasticNetCV
```

```python
# Define ratios and alphas to test
ratios = [0.1, 0.5, 0.7, 0.9, 0.95, 0.99, 1]
alphas = [0.001, 0.1, 1, 2, 10]

en_cv_model = ElasticNetCV(
    l1_ratio=ratios,
    alphas=alphas,
    cv=5,
    random_state=42
)

en_cv_model.fit(train_x, train_y)
```

```python
print("Best Alpha: ", en_cv_model.alpha_)
print("Best L1 Ratio: ", en_cv_model.l1_ratio_)
```

```python
# Final Evaluation
y_pred_cv = en_cv_model.predict(test_x)
r2 = r2_score(test_y, y_pred_cv)
print("Final R2 Score: ", r2)
```
