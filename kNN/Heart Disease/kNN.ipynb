### Load the libraries needed

```python
import pandas as pd
from sklearn.metrics import precision_score, accuracy_score, recall_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

### Load the Data

```python
# load dataset
df = pd.read_csv("heart.csv")
df.head()
```

### Split the data into **dependent** and **independent** variables

```python
# split the data into independent (features) and dependent (target) label sets
x = df.drop("target", axis = 1)
x.head()
```

```python
y = df.target
y.head()
```

### Split the data into training and testing part

```python
train_x, test_x, train_y, test_y = train_test_split(x, y, test_size=0.2, random_state = 42)
```

```python
len(train_x), len(test_x), len(train_y), len(test_y)
```

### Scale data

```python
scaler = StandardScaler()
train_x_scaled = scaler.fit_transform(train_x)
test_x_scaled = scaler.transform(test_x)
```

### Train the model

```python
# Create object
knn = KNeighborsClassifier(n_neighbors=3)
```

```python
# Pass the values to train the model
knn.fit(train_x_scaled, train_y)
```

```python
# predict values
y_pred = knn.predict(test_x_scaled)
y_pred
```

### Evaluation

```python
print("recall score: ", recall_score(test_y, y_pred))
print("accuracy score: ", accuracy_score(test_y, y_pred))
print("precision score: ", precision_score(test_y, y_pred))
```

### Cross Validation

```python
# Cross Validation for hyberparameter tunning using GridSearchCV
from sklearn.model_selection import GridSearchCV
```

```python
# CV
classifier = KNeighborsClassifier()
param_grid = {"n_neighbors": [3, 5, 7, 9]}

classifierCV = GridSearchCV(
    classifier,
    param_grid,
    cv=5,
    scoring="recall"
)

classifierCV.fit(train_x_scaled, train_y)
```

```python
# results
res = pd.DataFrame(classifierCV.cv_results_)
res
```

```python
# predict values
y_pred = classifierCV.predict(test_x_scaled)
y_pred
```

```python
# Evaluation
print("recall score: ", recall_score(test_y, y_pred))
print("accuracy score: ", accuracy_score(test_y, y_pred))
print("precision score: ", precision_score(test_y, y_pred))
```

### Pipeline

```python
# import pipeline
from sklearn.pipeline import Pipeline
```

```python
# create pipeline
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier())
])
```

```python
# redo CV

# prepend knn
param_grid = {"knn__n_neighbors": [3, 5, 7, 9]}

classifierCV = GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring="recall"
)

classifierCV.fit(train_x, train_y)
```

```python
# predict values
y_pred = classifierCV.predict(test_x)
y_pred
```

```python
# Evaluation
print("recall score: ", recall_score(test_y, y_pred))
print("accuracy score: ", accuracy_score(test_y, y_pred))
print("precision score: ", precision_score(test_y, y_pred))
```
