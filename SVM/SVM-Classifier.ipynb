### Load the libraries needed

```python
from sklearn.svm import SVC
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import datasets
```

### Load the data

```python
df = datasets.load_iris(as_frame=True).frame
df.head()
```

### Information about data

```python
# describe function to get numerical info about data
df.describe()
```

```python
# information about the data
df.info()
```

```python
# check number of null values
df.isnull().sum()
```

### Split the data into training and testing part

```python
# dependent variable
x = df.drop(columns=["target"])
x.head()
```

```python
# independent variable
y = df["target"]
y.head()
```

```python
# train test split
train_x, test_x, train_y, test_y = train_test_split(x, y, test_size=0.3, random_state = 42, stratify=y)
```

```python
len(train_x), len(test_x), len(train_y), len(test_y)
```

### Preprocessing data

```python
# import standard scaler
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
train_x_scaled = scaler.fit_transform(train_x)
test_x_scaled = scaler.transform(test_x)
```

### Create and Evaluate Model

```python
# create object
svc = SVC()
```

```python
# create model
svc.fit(train_x_scaled, train_y)
```

```python
# predict
y_pred = svc.predict(test_x_scaled)
y_pred
```

```python
# Evaluate
from sklearn.metrics import accuracy_score, classification_report

print("accuracy: ", accuracy_score(test_y, y_pred))
print("classification: ", classification_report(test_y, y_pred))
```

```python
# Linear kernel

# create model
svc = SVC(kernel="linear")
svc.fit(train_x_scaled, train_y)

# predict
y_pred = svc.predict(test_x_scaled)

# Evaluate
print("accuracy: ", accuracy_score(test_y, y_pred))
```

```python
# poly kernel

# create model
svc = SVC(kernel="poly")
svc.fit(train_x_scaled, train_y)

# predict
y_pred = svc.predict(test_x_scaled)

# Evaluate
print("accuracy: ", accuracy_score(test_y, y_pred))
```

```python
# sigmoid kernel

# create model
svc = SVC(kernel="sigmoid")
svc.fit(train_x_scaled, train_y)

# predict
y_pred = svc.predict(test_x_scaled)

# Evaluate
print("accuracy: ", accuracy_score(test_y, y_pred))
```

### Hyper parameter tuneing

```python
c_vals = [0.5, 1, 2, 3, 4, 5]

for c_val in c_vals:
    # create model
    svc = SVC(kernel="rbf", C=c_val)
    svc.fit(train_x_scaled, train_y)
    
    # predict
    y_pred = svc.predict(test_x_scaled)
    
    # Evaluate
    print("accuracy:", accuracy_score(test_y, y_pred), "for C :", c_val)
```
